# **Particle Swarm Optimization**(PSO)

PSO is a population-based optimization algorithm that is inspired by the behavior of bird flocking or fish schooling.



```python
from pso import *

Sphere = Problem(lambda x: np.sum(x**2), 30, -100, 100)
```

```python
pso = PSO(Sphere, 1200, 1000, 0.7, 0.9, 1.2, 0.7)
best_pos = pso.optimize()
print(best_pos)
print(Sphere(best_pos))
```
